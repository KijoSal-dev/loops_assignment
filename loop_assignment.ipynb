```python
#1. Write a loop that prints the first 5 multiples of 3

for i in range(1,6):
    print(i*3)
```

```python
#2. Given a list of scores, write a loop that counts how many scores are greater than 50

scores = [45, 67, 89, 34, 50, 72, 18, 95, 100, 29]

for i in scores:
    if i > 50:
      print (i)
```

```python
#3. Write a program using a while loop that prints numbers from 10 down to 1.

for i in range(10,0,-1):
  print(i)
```

```python
#4. Given a string, write a loop that prints each character one by one.

text = "techstars"

for char in text:
  print(char)
```

```python
#5. Write a loop that stops when it encounters the number 7 in a list of numbers

numbers = [105, 82, 53, 74, 35, 67, 8, 90, 10, 7, 13, 15, 20 , 45, 100 ]

for i in numbers:
  if i == 7:
      break
  print (i)
```

```python
#6. Write a program that prints numbers from 1 to 10 using a loop

count = 1

while count <= 10:
  print(count)
  count += 1
```

```python
#7. Given a list of names, write a loop that prints each name.

names = ["Will", "Jamie", "Liz", "Vecna"]

for i in names:
  print(i)
```

```python
#8. Write a program that keeps asking the user for a number until the user enters 0.

num = int(input("Enter a number: "))

while num != 0:
    num = int(input("Enter a number: "))
```

```python
#9. Given a list of numbers, write a loop that prints only the even numbers.

numbers = [15, 20, 30, 77, 99, 55, 60, 100, 40]

for i in numbers:
  if i % 2 == 0:
    print(i)
```

```python
#10. What will be the output of a loop that starts at 1 and increases by 2 until it reaches 10?

count = 1

while count <= 10:
    print(count)
    count +=2
```
